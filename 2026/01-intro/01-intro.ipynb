##### 6. Median value of horsepower

```python
# Check the median of horsepower
median = df["horsepower"].median()

# Check the mode of horsepower
mode = df["horsepower"].mode()[0]

# Fill the missing values with fillna in horsepower
df_copy = df.copy()
df_copy["horsepower"] = df_copy["horsepower"].fillna(mode)

# Get the new median
median_new = df_copy["horsepower"].median()

print(f"The old median is {median} and the new one is {median_new}")
```

##### 7. Sum of weights

```python
# Select only two columns
columns = ["vehicle_weight", "model_year"]

# Create a new dataframe to work with
df_asia = df[df["origin"] == "Asia"][columns].head(7)

# Get the underlying NumPy array
X = df_asia.values

# Matrix multiplication to create a square
X_T = X.T
XTX = X_T @ X

# Invert XTX where we need a square, because otherwise the invert isn't possible
XTX_inv = np.linalg.inv(XTX)

# Create an array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# Multiply the inverse with transpose of X, and then by y
w = XTX_inv @ X_T @ y
w.sum()
```
